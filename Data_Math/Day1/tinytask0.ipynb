Here are 10 years of anomaly values. They're a simplified illustration, not the real dataset:

```python
years = [2014, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023]
anoms = [0.75, 0.90, 1.01, 0.92, 0.85, 0.98, 1.01, 0.85, 0.89, 1.17]
```

Predict first, before writing any code: out of the 9 year-to-year changes, how many go up?

4-5

Then write code that counts how many years were warmer than the year before, and how many were cooler.

```python
prev = None
warmer = 0
cooler = 0
for year, anom in zip(years, anoms):
    if prev is not None:
        if anom > prev:
            warmer += 1
        elif prev > anom:
            cooler += 1
    prev = anom
print(f'warmer: {warmer} cooler: {cooler}')
```

Last, print how much warmer 2023 is than 2014.

```python
year_anom = dict(zip(years, anoms))
print(year_anom[2023] - year_anom[2014])
```
